### Module 1.2: System Calls & The Gatekeepers of Ring 0

Your application runs in CPU **Ring 3 (User Mode)**. It cannot write directly to the hardware, modify memory maps of other processes, or send packets out of the network card. Only the kernel, running in **Ring 0 (Kernel Mode)**, has the authorization to perform these tasks.

When a program in user space wants to do something privileged (like reading a file from disk or printing text to a terminal), it executes a **System Call (syscall)**. This triggers a software interrupt, switching the CPU execution mode from Ring 3 to Ring 0. The kernel handles the request, and then drops the privilege mode back down to Ring 3 before returning control to the user program.

#### Common System Calls include:
*   `open()`, `read()`, `write()`, `close()` — File I/O operations
*   `fork()`, `execve()` — Creating and replacing execution processes
*   `brk()`, `sbrk()`, `mmap()` — Allocating and mapping system memory

#### Tracking Syscalls in Action
On Linux, we can trace every single system call a process makes using a utility called `strace`.

### Module 1.3: Standard I/O Streams & Redirection Mechanics

When any process is spawned, the kernel automatically spins it up with three default **File Descriptors (FD)**:
*   **`0` (Standard Input / `stdin`):** Typically mapped to your keyboard/terminal entry.
*   **`1` (Standard Output / `stdout`):** Mapped to your screen or console output.
*   **`2` (Standard Error / `stderr`):** Also mapped to your screen, used specifically for error logs to keep them separate from normal output.

#### The Mechanics of Redirection
When we use shell redirection operators like `>` or `2>`, we are directing the shell to swap these file descriptors *before* executing the program.
*   `command > file.txt`: Redirects File Descriptor `1` (`stdout`) to a file.
*   `command 2> error.log`: Redirects File Descriptor `2` (`stderr`) to a file.
*   `command 2>&1`: Merges File Descriptor `2` directly into File Descriptor `1` so both errors and normal output flow into the same stream.

### Module 1.4: Process Inheritance — Shell Variables vs. Environment Variables

When working in a shell, you can define variables. However, there is a vital distinction between a **Local Shell Variable** and an **Environment Variable**:

*   **Local Shell Variable:** Only exists within the current active instance of the shell interpreter. If that shell spawns a child process (like executing a Python script or a compiled binary), the child process *does not* inherit or see this variable.
*   **Environment Variable:** Marked with the `export` attribute. When the shell spawns any child process, the child process inherits a complete copy of these environment variables.

Let's test this inheritance behavior using Python to programmatically verify what variables are passed from our shell down to a child process.

````python
# 1. To persistently change the environment of our main notebook process, we must modify Python's memory mapping of the environment table.
import os

# Let's set an environment variable within our current process
os.environ["ENV_VAR"] = "I am exported persistently from the main process memory!"

# 2. We can also use Colab's persistent magic command, which sets it in the running process environment
%env MAGIC_VAR=I am a persistent magic variable

# 3. Now let's execute a shell command as a child process.
# Since the variables are in the parent process's environment table, the child process WILL inherit them!
print("=== Checking what Python sees in its memory ===")
print("Env Var in Python:", os.environ.get("ENV_VAR"))
print("Magic Var in Python:", os.environ.get("MAGIC_VAR"))

print("\n=== Checking what a child shell process (!) inherits ===")
!echo "Child shell sees: $ENV_VAR"
!echo "Child shell sees: $MAGIC_VAR"
````

### Module 1.5: Shell Initialization Files (Login vs. Non-Login)

Every time you access a terminal, several shell scripts run automatically to configure your environment, aliases, and paths. The exact files that execute depend on how the shell is initiated:

1.  **Interactive Login Shell** (e.g., Logging in via SSH or console):
    *   First reads system-wide configurations: `/etc/profile`
    *   Then searches for your user-specific configuration files in order, executing **only the first one** it finds:
        1. `~/.bash_profile`
        2. `~/.bash_login`
        3. `~/.profile`

2.  **Interactive Non-Login Shell** (e.g., Opening a new terminal tab after logging in, or running nested shells):
    *   Reads system-wide settings: `/etc/bash.bashrc`
    *   Reads user-specific settings: `~/.bashrc`

### Module 1.6: Injecting Settings into Startup Scripts

When we edit a startup configuration file like `~/.bashrc`, the changes do not immediately apply to your active shell process. This is because the configuration is only parsed once when the shell is initiated.

To apply the changes immediately, you have two choices:
1.  **`source ~/.bashrc`** (or use the `.` shorthand): Instructs the current active shell process to parse and execute the script inside the active memory space.
2.  **Spawn a new subshell**: Let the initialization run naturally on process instantiation.

Let's run a test where we append an alias and an environment variable to `~/.bashrc`, then execute a non-login bash subshell to see if it configures itself automatically.

````python
# 1. Explicitly source the configuration file inside a nested execution context
print("=== Method 1: Sourcing ~/.bashrc explicitly inside our command ===")
!bash -c 'source ~/.bashrc && echo "DEMO_STARTUP_VAR: $DEMO_STARTUP_VAR"'

# 2. Force an interactive subshell execution using the -i flag
print("\n=== Method 2: Forcing an Interactive Shell (-i) ===")
!bash -ic 'echo "DEMO_STARTUP_VAR: $DEMO_STARTUP_VAR"'

# 3. Test if our custom command alias 'sysstatus' works in the interactive shell environment too!
print("\n=== Testing our custom alias 'sysstatus' inside an Interactive Shell ===")
!bash -ic 'sysstatus'
````
